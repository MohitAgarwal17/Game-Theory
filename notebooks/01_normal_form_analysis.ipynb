## 1. Problem Overview
When two autonomous vehicles arrive simultaneously at an unsignalized intersection, each agent faces a strategic choice between **Proceed (Go)** and **Yield**.
This interaction is formulated as a simultaneous normal-form game (Chicken / Hawk-Dove Game).

### Payoff Matrix
$$\begin{array}{c|cc}
A \backslash B & \text{Go} & \text{Yield} \\
\hline
\text{Go} & (-100, -100) & (10, 2) \\
\text{Yield} & (2, 10) & (0, 0)
\end{array}$$

## 2. Solving the Symmetric Chicken Game
We initialize the NormalFormGame object and compute pure and mixed-strategy Nash equilibria.


```python
payoff_a, payoff_b = chicken_game()
game = NormalFormGame(
    payoff_a, payoff_b,
    strategy_labels_a=["Go", "Yield"],
    strategy_labels_b=["Go", "Yield"],
    player_names=("Vehicle A", "Vehicle B")
)
print("Normal-Form Game Specification:")
print(game)

solver = NashSolver()
equilibria = solver.solve(game)

print(f"\nFound {len(equilibria)} Nash Equilibria:")
for i, eq in enumerate(equilibria, 1):
    ptype = "Pure" if eq.is_pure else "Mixed"
    print(f"  EQ {i} ({ptype}): {eq.strategy_label(game)} | Payoffs: ({eq.payoff_a:.2f}, {eq.payoff_b:.2f}) | Pareto Optimal: {eq.is_pareto_optimal}")
```

## 3. Best-Response Correspondences & Mixed-Strategy Indifference
For Player A to be indifferent between Go and Yield:
$$E[u_A(\text{Go})] = q(-100) + (1-q)(10) = 10 - 110q$$
$$E[u_A(\text{Yield})] = q(2) + (1-q)(0) = 2q$$
$$10 - 110q = 2q \implies 112q = 10 \implies q^* = \frac{10}{112} \approx 0.0893$$

Under mixed strategy, the collision probability is:
$$P(\text{Collision}) = p^* \times q^* = (0.0893)^2 \approx 0.80\%$$


```python
# Plot payoff matrix heatmap and best response correspondence
fig_heatmap = plot_payoff_matrix(game, title="Chicken Game Payoffs")
plt.show()

fig_br = plot_best_response(game)
plt.show()
```

## 4. Priority Asymmetry (Main Road vs. Side Road)
In real-world traffic, one vehicle often approaches on an arterial highway, conferring priority bonus.


```python
asym_a, asym_b = asymmetric_intersection(priority_bonus=5.0)
asym_game = NormalFormGame(
    asym_a, asym_b,
    strategy_labels_a=["Go", "Yield"],
    strategy_labels_b=["Go", "Yield"],
    player_names=("Main Road", "Side Road")
)
asym_eqs = solver.solve(asym_game)

print("Asymmetric Priority Game Equilibria:")
for eq in asym_eqs:
    print(f"  {eq.strategy_label(asym_game)} | Payoffs: ({eq.payoff_a:.2f}, {eq.payoff_b:.2f})")
```

## 5. Conclusions
- Pure NE (Go, Yield) and (Yield, Go) are Pareto optimal but present an unresolvable coordination problem without signaling or sequence.
- Mixed NE induces a non-zero collision risk ($0.80\%$) and drastically diminishes social welfare ($0.36$ vs $12.0$).
- This highlights the fundamental necessity of sequential or mechanism-based right-of-way management.

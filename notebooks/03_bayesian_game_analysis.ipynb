## 1. Introduction
In mixed-autonomy traffic, an Autonomous Vehicle (Player 1) must interact with human-driven vehicles whose internal driver temperament is unobservable.
We model the human driver as belonging to one of two types drawn by Nature:
- **Aggressive ($\theta_{\text{agg}}$):** High valuation on speed, dominant strategy to Go.
- **Cautious ($\theta_{\text{caut}}$):** Safety-conscious, yields when challenged.

Prior distribution: $P(\theta_{\text{agg}}) = p, \quad P(\theta_{\text{caut}}) = 1 - p$.


```python
import os
import sys
import numpy as np
import matplotlib.pyplot as plt

# Add root directory to path
sys.path.insert(0, os.path.abspath('..'))

from src.games.bayesian_game import BayesianGame
from src.solvers.bayesian_solver import BayesianSolver
```

## 2. Solving Bayesian Nash Equilibrium (BNE)


```python
bgame = BayesianGame.create_av_vs_human(prob_aggressive=0.35)
solver = BayesianSolver()
res = solver.solve(bgame)

print(f"Bayesian Game Solved:")
print(f"  AV Yield Probability: {res.av_yield_rate:.1%}")
print(f"  Expected Collision Rate: {res.expected_collision_rate:.2%}")
print(f"  Expected Payoffs: {res.expected_payoffs}")
```

## 3. Sensitivity Analysis across Human Driver Aggressiveness


```python
probs = np.linspace(0.05, 0.95, 20)
av_yield_probs = []
expected_collision_probs = []

for p in probs:
    bg = BayesianGame.create_av_vs_human(prob_aggressive=p)
    sol = solver.solve(bg)
    av_yield_probs.append(sol.av_yield_rate)
    expected_collision_probs.append(sol.expected_collision_rate)

plt.figure(figsize=(8, 4.5))
plt.plot(probs, av_yield_probs, 'o-', color='#00d4ff', label='AV Yield Rate')
plt.plot(probs, expected_collision_probs, 's--', color='#e74c3c', label='Expected Collision Risk')
plt.xlabel('Prior Probability of Aggressive Driver P(θ_agg)')
plt.ylabel('Probability')
plt.title('AV Bayesian Equilibrium Strategy Sensitivity')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. Conclusions
- As the perceived likelihood of an aggressive driver increases, the AV rationally shifts towards a defensive yielding strategy.
- When $P(\theta_{\text{agg}}) \ge 0.10$, the catastrophic penalty of collision dominates, driving the AV's optimal response to **Yield**.
- Defensive driving in autonomous systems is an equilibrium outcome of incomplete information games.

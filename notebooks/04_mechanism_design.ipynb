## 1. Introduction
When multiple vehicles ($N \ge 4$) converge at a 4-way intersection, centralized or decentralized market mechanisms can dynamically allocate right-of-way based on private urgency valuations.

### Mechanisms Evaluated
1. **Vickrey Auction (Second-Price Sealed-Bid):**
   - Winner: highest bidder.
   - Payment: second-highest bid.
   - Incentive Compatibility: truth-telling is weakly dominant ($b_i^* = v_i$).
2. **First-Price Sealed-Bid Auction:**
   - Winner: highest bidder, pays own bid ($b_i < v_i$ shaded bidding).
3. **Priority Borda Count Voting:**
   - Non-monetary civic voting taking into account occupancy, transit status, and emergency preemption.

## 3. Incentive Compatibility & Dominant Strategy Truth-Telling
In a second-price auction, any unilateral deviation from true valuation $v_i$ yields weakly lower payoff:
$$\pi_i(v_i, b_{-i}) \ge \pi_i(b_i, b_{-i}) \quad \forall b_i \neq v_i$$


```python
vickrey = VickreyAuction()
bids = {'Car_1': 14.5, 'Car_2': 18.2, 'Car_3': 9.0, 'Car_4': 12.1}
result = vickrey.run(bids)

print("Single Vickrey Urgency Auction Execution:")
print(f"  Crossing Order: {result.crossing_order}")
print(f"  Winner: {result.winner_id} (Bid: {result.winning_bid})")
print(f"  Payment (Second-Highest): {result.payment:.2f}")
print(f"  Winner Surplus: {result.winner_surplus:.2f}")
```

## 4. Priority Voting with Emergency Preemption
Non-monetary coordination mechanism where civic vehicles, buses, and ambulances cast weighted ballots.


```python
voters = [
    VehicleVoter("Ambulance", is_emergency=True, occupancy=2, wait_time=2.0),
    VehicleVoter("City_Bus", is_emergency=False, occupancy=35, wait_time=5.0),
    VehicleVoter("Sedan_A", is_emergency=False, occupancy=1, wait_time=8.0),
    VehicleVoter("Sedan_B", is_emergency=False, occupancy=1, wait_time=3.0),
]

voting_mech = PriorityVotingMechanism(method="borda")
order = voting_mech.determine_order(voters)
print("Priority Voting Crossing Order:")
for rank, v_id in enumerate(order, 1):
    print(f"  Rank {rank}: {v_id}")
```

## 1. Introduction
To resolve the coordination breakdown of simultaneous play, vehicles naturally establish sequential ordering based on **Time-to-Stop-Line (TTSL)**.
The closer vehicle acts as the **Leader**, choosing an action first; the further vehicle acts as the **Follower**, observing the leader's commitment and best-responding.

### Backward Induction
1. At follower nodes:
   - If Leader plays **Go**, Follower chooses between Go $(-100)$ and Yield $(2) \implies$ Follower plays **Yield**.
   - If Leader plays **Yield**, Follower chooses between Go $(10)$ and Yield $(0) \implies$ Follower plays **Go**.
2. At leader root node:
   - Anticipating Follower's best responses, Leader compares Go $(10)$ vs Yield $(2) \implies$ Leader chooses **Go**.


```python
import os
import sys
import numpy as np

# Add root directory to path
sys.path.insert(0, os.path.abspath('..'))

from src.games.normal_form import NormalFormGame
from src.games.extensive_form import ExtensiveFormGame
from src.solvers.stackelberg_solver import StackelbergSolver
from src.utils.payoff_matrices import chicken_game
```

## 2. Solving Stackelberg Equilibrium via Backward Induction


```python
payoff_a, payoff_b = chicken_game()
game = NormalFormGame(payoff_a, payoff_b, strategy_labels_a=["Go", "Yield"], strategy_labels_b=["Go", "Yield"])

solver = StackelbergSolver()
stack_eq = solver.solve_from_normal_form(game)

print("Stackelberg Leader-Follower Equilibrium:")
print(f"  Leader Action: {'Go' if stack_eq.leader_action == 0 else 'Yield'}")
print(f"  Follower Response: Yield")
print(f"  Leader Payoff: {stack_eq.leader_payoff}")
print(f"  Follower Payoff: {stack_eq.follower_payoff}")
print(f"  Equilibrium Path: {stack_eq.outcome}")
```

## 3. First-Mover Advantage Analysis
We compare payoffs when Vehicle A leads versus when Vehicle B leads.


```python
adv = solver.compare_leader_advantage(game)
print("First-Mover Advantage Comparison:")
print(f"  When A Leads: A gets {adv['a_leads']['leader_payoff']}, B gets {adv['a_leads']['follower_payoff']}")
print(f"  When B Leads: B gets {adv['b_leads']['leader_payoff']}, A gets {adv['b_leads']['follower_payoff']}")
print(f"  First-Mover Surplus: {adv['first_mover_advantage']}")
```

## 4. Extensive-Form Tree Representation
Using the `ExtensiveFormGame` class to verify subgame-perfect equilibrium directly on game trees.


```python
tree = ExtensiveFormGame.from_normal_form(game, leader_idx=0)
print(f"Extensive Game Tree created with {len(tree.nodes)} nodes.")
spe = solver.solve_tree(tree)
print(f"SPE Action Profile: {spe}")
```
